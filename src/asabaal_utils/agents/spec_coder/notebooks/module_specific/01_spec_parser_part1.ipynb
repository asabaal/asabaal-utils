```python
# Cell 1: Setup and Dependencies
import yaml
from dataclasses import dataclass
from typing import List, Dict, Any, Optional
from pathlib import Path

print("✅ Module 1: SpecParser - Foundation Setup Complete")
print("Dependencies imported:")
print("  • yaml - For parsing YAML specification files")
print("  • dataclasses - For structured data representation")
print("  • typing - For type hints and better code understanding")
print("  • pathlib - For modern file path handling")
```

```python
# Cell 2: Data Classes - The Building Blocks
@dataclass
class ValidationCriteria:
    """Represents validation criteria for a requirement."""
    type: str  # 'unit', 'integration', etc.
    file: str
    target: str

@dataclass
class FunctionInterface:
    """Represents a function interface specification."""
    function: str
    parameters: Dict[str, str]  # parameter_name: type
    returns: str
    example: Optional[Dict[str, Any]] = None

@dataclass
class Requirement:
    """Represents a single requirement in the specification."""
    id: str
    title: str
    description: str
    interface: Optional[FunctionInterface] = None
    validation: Optional[List[ValidationCriteria]] = None

@dataclass
class OpenSpec:
    """Represents a complete OpenSpec specification."""
    spec_id: str
    title: str
    version: str
    requirements: List[Requirement]
    raw_data: Dict[str, Any]

print("✅ Data classes defined:")
print("  • ValidationCriteria - How to validate requirements")
print("  • FunctionInterface - Function signatures and types")
print("  • Requirement - Individual requirement specification")
print("  • OpenSpec - Complete specification container")
```

```python
# Cell 3: SpecParser Class - Constructor
class SpecParser:
    """Parser for OpenSpec YAML specification files."""
    
    def __init__(self):
        self.specs: Dict[str, OpenSpec] = {}

# Test the constructor
parser = SpecParser()
print(f"✅ SpecParser created: {type(parser)}")
print(f"Initial specs dictionary: {parser.specs}")
print(f"Specs dictionary type: {type(parser.specs)}")
print("\nConstructor breakdown:")
print("  • self.specs = {} - Creates empty dictionary to store parsed specs")
print("  • Dict[str, OpenSpec] - Maps spec_id to OpenSpec objects")
```

```python
# Cell 4: parse_file Method - Reading YAML Files
def add_parse_file_method(parser_class):
    def parse_file(self, file_path: Path) -> OpenSpec:
        """Parse a single OpenSpec YAML file."""
        with open(file_path, 'r', encoding='utf-8') as f:
            data = yaml.safe_load(f)
        
        return self._parse_data(data, str(file_path))
    
    parser_class.parse_file = parse_file

# Add the method to our parser
add_parse_file_method(SpecParser)

print("✅ parse_file method added to SpecParser")
print("\nMethod breakdown:")
print("  • file_path: Path - Path to YAML specification file")
print("  • open(file_path, 'r', encoding='utf-8') - Safe file opening with UTF-8")
print("  • yaml.safe_load(f) - Parse YAML safely (no code execution)")
print("  • self._parse_data(data, str(file_path)) - Delegate to internal parser")
print("  • Returns: OpenSpec - Parsed specification object")
```

```python
# Cell 5: parse_string Method - Parsing YAML from Strings
def add_parse_string_method(parser_class):
    def parse_string(self, yaml_content: str, source: str = "<string>") -> OpenSpec:
        """Parse OpenSpec YAML from a string."""
        data = yaml.safe_load(yaml_content)
        return self._parse_data(data, source)
    
    parser_class.parse_string = parse_string

# Add the method to our parser
add_parse_string_method(SpecParser)

print("✅ parse_string method added to SpecParser")
print("\nMethod breakdown:")
print("  • yaml_content: str - YAML content as string")
print("  • source: str = '<string>' - Source identifier for error reporting")
print("  • yaml.safe_load(yaml_content) - Parse YAML from string")
print("  • self._parse_data(data, source) - Same internal parser as file method")
print("  • Returns: OpenSpec - Parsed specification object")
```

```python
# Cell 6: _parse_data Method - Core Parsing Logic
def add_parse_data_method(parser_class):
    def _parse_data(self, data: Dict[str, Any], source: str) -> OpenSpec:
        """Parse raw YAML data into OpenSpec object."""
        # Extract basic metadata
        spec_id = data.get('spec_id', 'unknown')
        title = data.get('title', 'Untitled Specification')
        version = data.get('version', '0.1.0')
        
        # Parse requirements
        requirements = []
        for req_data in data.get('requirements', []):
            requirement = self._parse_requirement(req_data)
            requirements.append(requirement)
        
        return OpenSpec(
            spec_id=spec_id,
            title=title,
            version=version,
            requirements=requirements,
            raw_data=data
        )
    
    parser_class._parse_data = _parse_data

# Add the method to our parser
add_parse_data_method(SpecParser)

print("✅ _parse_data method added to SpecParser")
print("\nMethod breakdown:")
print("  • data.get('spec_id', 'unknown') - Extract spec ID with fallback")
print("  • data.get('title', 'Untitled') - Extract title with fallback")
print("  • data.get('version', '0.1.0') - Extract version with fallback")
print("  • data.get('requirements', []) - Get requirements list (empty if missing)")
print("  • for req_data in requirements: - Loop through each requirement")
print("  • self._parse_requirement(req_data) - Parse individual requirement")
print("  • OpenSpec(...) - Create final OpenSpec object with all data")
```

```python
# Cell 7: _parse_requirement Method - Individual Requirement Parsing
def add_parse_requirement_method(parser_class):
    def _parse_requirement(self, req_data: Dict[str, Any]) -> Requirement:
        """Parse a single requirement from YAML data."""
        req_id = req_data.get('id', 'unknown')
        title = req_data.get('title', 'Untitled Requirement')
        description = req_data.get('description', '')
        
        # Parse interface if present
        interface = None
        if 'interface' in req_data:
            interface_data = req_data['interface']
            interface = FunctionInterface(
                function=interface_data.get('function', ''),
                parameters=interface_data.get('parameters', {}),
                returns=interface_data.get('returns', ''),
                example=interface_data.get('example')
            )
        
        # Parse validation criteria
        validation = []
        validation_data = req_data.get('validation', [])
        
        # Handle both string and list formats for validation
        if isinstance(validation_data, str):
            # Simple string validation
            criteria = ValidationCriteria(
                type='unit',
                file='',
                target=validation_data
            )
            validation.append(criteria)
        elif isinstance(validation_data, list):
            # List of validation criteria
            for val_data in validation_data:
                criteria = ValidationCriteria(
                    type=val_data.get('type', 'unit'),
                    file=val_data.get('file', ''),
                    target=val_data.get('target', 'pass')
                )
                validation.append(criteria)
        
        return Requirement(
            id=req_id,
            title=title,
            description=description,
            interface=interface,
            validation=validation
        )
    
    parser_class._parse_requirement = _parse_requirement

# Add the method to our parser
add_parse_requirement_method(SpecParser)

print("✅ _parse_requirement method added to SpecParser")
print("\nMethod breakdown:")
print("  • Extract basic fields: id, title, description with fallbacks")
print("  • Parse interface if present - creates FunctionInterface object")
print("  • Parse validation criteria - handles both string and list formats")
print("  • String validation: creates single ValidationCriteria with type='unit'")
print("  • List validation: creates ValidationCriteria for each item")
print("  • Return Requirement object with all parsed data")
```

```python
# Cell 8: Test the Core Parsing Functionality
# Create a sample YAML specification
sample_yaml = """
spec_id: "test_spec_001"
title: "Test Specification"
version: "1.0.0"
requirements:
  - id: "REQ-001"
    title: "Calculate Sum"
    description: "Function to calculate sum of two numbers"
    interface:
      function: "calculate_sum"
      parameters:
        a: "int"
        b: "int"
      returns: "int"
      example:
        a: 5
        b: 3
        result: 8
    validation:
      - type: "unit"
        file: "test_calculate_sum.py"
        target: "test_basic_addition"
"""

# Test parsing
parser = SpecParser()
parsed_spec = parser.parse_string(sample_yaml, "test_spec")

print(f"✅ Successfully parsed specification:")
print(f"  • Spec ID: {parsed_spec.spec_id}")
print(f"  • Title: {parsed_spec.title}")
print(f"  • Version: {parsed_spec.version}")
print(f"  • Requirements count: {len(parsed_spec.requirements)}")

if parsed_spec.requirements:
    req = parsed_spec.requirements[0]
    print(f"\nFirst requirement:")
    print(f"  • ID: {req.id}")
    print(f"  • Title: {req.title}")
    print(f"  • Description: {req.description}")
    if req.interface:
        print(f"  • Function: {req.interface.function}({', '.join(req.interface.parameters.keys())}) -> {req.interface.returns}")
    if req.validation:
        print(f"  • Validation: {len(req.validation)} criteria")
```

```python
# Cell 9: format_requirements_for_prompt Method - AI Integration
def add_format_requirements_method(parser_class):
    def format_requirements_for_prompt(self, spec: OpenSpec) -> str:
        """Format requirements for inclusion in AI prompts."""
        formatted = []
        for req in spec.requirements:
            req_text = f"- {req.id}: {req.title}\n"
            req_text += f"  Description: {req.description}\n"
            if req.validation:
                req_text += f"  Validation: {', '.join([f'{v.type} test in {v.file}' for v in req.validation])}\n"
            formatted.append(req_text)
        
        return '\n'.join(formatted)
    
    parser_class.format_requirements_for_prompt = format_requirements_for_prompt

# Add the method to our parser
add_format_requirements_method(SpecParser)

# Test the method
parser = SpecParser()
parsed_spec = parser.parse_string(sample_yaml, "test_spec")
formatted_reqs = parser.format_requirements_for_prompt(parsed_spec)

print("✅ format_requirements_for_prompt method added and tested")
print("\nFormatted requirements for AI prompt:")
print(formatted_reqs)

print("\nMethod breakdown:")
print("  • Loop through each requirement in the specification")
print("  • Format: '- ID: Title' with newline")
print("  • Add description with proper indentation")
print("  • Add validation criteria if present")
print("  • Join all formatted requirements with newlines")
```

```python
# Cell 10: format_function_signature Method - Code Generation Helper
def add_format_function_signature_method(parser_class):
    def format_function_signature(self, requirement: Requirement) -> str:
        """Format a function signature from the requirement interface."""
        if not requirement.interface:
            # Fallback to old format if no interface
            return f"def {requirement.title.lower().replace(' ', '_')}():\n    \"\"\"TODO {requirement.id}: {requirement.description}.\"\"\"\n    pass"
        
        interface = requirement.interface
        
        # Format parameters
        params = []
        for param_name, param_type in interface.parameters.items():
            params.append(f"{param_name}: {param_type}")
        
        param_str = ", ".join(params)
        
        return f"def {interface.function}({param_str}) -> {interface.returns}:\n    \"\"\"TODO {requirement.id}: {requirement.description}.\"\"\"\n    pass"
    
    parser_class.format_function_signature = format_function_signature

# Add the method to our parser
add_format_function_signature_method(SpecParser)

# Test the method
parser = SpecParser()
parsed_spec = parser.parse_string(sample_yaml, "test_spec")
req = parsed_spec.requirements[0]
function_sig = parser.format_function_signature(req)

print("✅ format_function_signature method added and tested")
print("\nGenerated function signature:")
print(function_sig)

print("\nMethod breakdown:")
print("  • Check if requirement has interface")
print("  • Fallback: create function from title if no interface")
print("  • Format parameters as 'name: type' pairs")
print("  • Join parameters with commas")
print("  • Create complete function signature with docstring")
```

```python
# Cell 11: Additional Utility Methods
def add_utility_methods(parser_class):
    def get_requirement_by_id(self, spec: OpenSpec, req_id: str) -> Optional[Requirement]:
        """Get a specific requirement by its ID."""
        for req in spec.requirements:
            if req.id == req_id:
                return req
        return None
    
    def validate_spec(self, spec: OpenSpec) -> List[str]:
        """Validate the specification and return any issues."""
        issues = []
        
        if not spec.spec_id:
            issues.append("Missing spec_id")
        
        if not spec.title:
            issues.append("Missing title")
        
        if not spec.requirements:
            issues.append("No requirements found")
        
        for i, req in enumerate(spec.requirements):
            if not req.id:
                issues.append(f"Requirement {i+1} missing ID")
            if not req.title:
                issues.append(f"Requirement {i+1} missing title")
            if not req.description:
                issues.append(f"Requirement {i+1} missing description")
        
        return issues
    
    parser_class.get_requirement_by_id = get_requirement_by_id
    parser_class.validate_spec = validate_spec

# Add utility methods
add_utility_methods(SpecParser)

# Test the methods
parser = SpecParser()
parsed_spec = parser.parse_string(sample_yaml, "test_spec")

# Test get_requirement_by_id
found_req = parser.get_requirement_by_id(parsed_spec, "REQ-001")
print(f"✅ Found requirement: {found_req.title if found_req else 'None'}")

# Test validate_spec
issues = parser.validate_spec(parsed_spec)
print(f"✅ Validation issues: {len(issues)}")
if issues:
    for issue in issues:
        print(f"  • {issue}")
else:
    print("  • No issues found - specification is valid!")
```
