```python
# Cell 2: Mock Dependencies for Demonstration
class MockTestSummarizer:
    """Mock TestSummarizer for demonstration purposes."""
    def __init__(self, model_name: str, spec_file: Optional[Path] = None, source_file: Optional[Path] = None):
        self.model_name = model_name
        self.spec_file = spec_file
        self.source_file = source_file
    
    def summarize_test_file(self, parsed_data: Dict[str, Any]) -> Dict[str, Any]:
        """Mock AI summarization of test file."""
        return {
            **parsed_data,
            'behaviors': [
                {
                    'function': 'test_example',
                    'behavior': 'Validates basic functionality',
                    'confidence': 0.9,
                    'coverage': ['unit', 'integration']
                }
            ],
            'coverage_analysis': {
                'total_functions': len(parsed_data.get('functions', [])),
                'behaviors_extracted': 1,
                'quality_score': 0.85
            },
            'recommendations': [
                'Add more edge case tests',
                'Include error handling tests'
            ]
        }

def mock_parse_test_file(test_file_path: Path) -> Dict[str, Any]:
    """Mock AST parsing of test file."""
    if not test_file_path.exists():
        return {'error': f'File not found: {test_file_path}'}
    
    # Mock parsed data structure
    return {
        'file_path': str(test_file_path),
        'functions': [
            {
                'name': 'test_example',
                'line': 1,
                'args': [],
                'docstring': 'Example test function',
                'decorators': [],
                'assertions': ['assert True']
            }
        ],
        'classes': [],
        'imports': ['import pytest'],
        'ast_valid': True
    }

print("✅ Mock dependencies defined:")
print("  • MockTestSummarizer - Simulates AI-powered test analysis")
print("  • mock_parse_test_file - Simulates AST parsing of test files")
print("  • Both provide realistic data structures for demonstration")
```

```python
# Cell 4: analyze_single_file Method - Core Analysis Function
def add_analyze_single_file_method(analyzer_class):
    def analyze_single_file(self, test_file_path: Path) -> Dict[str, Any]:
        """
        Analyze a single test file using both AST parsing and AI summarization.
        
        This method performs comprehensive analysis of a test file by first parsing
        the AST to extract structural information, then using AI to understand
        the semantic meaning and behavior of each test.
        """
        logger.info(f"Analyzing test file: {test_file_path}")
        
        # Step 1: Parse the test file with AST
        parsed_data = mock_parse_test_file(test_file_path)
        
        if "error" in parsed_data:
            logger.error(f"Failed to parse {test_file_path}: {parsed_data['error']}")
            return parsed_data
        
        # Step 2: Summarize with AI
        summarized_data = self.summarizer.summarize_test_file(parsed_data)
        
        return summarized_data
    
    analyzer_class.analyze_single_file = analyze_single_file

# Add the method to our analyzer
add_analyze_single_file_method(TestAnalyzer)

# Create a test file for analysis
test_file_content = '''
import pytest

def test_example_function():
    """Example test function."""
    result = 1 + 1
    assert result == 2

def test_another_case():
    """Another test case."""
    data = [1, 2, 3]
    assert len(data) == 3
'''

test_file_path = Path("test_example.py")
with open(test_file_path, 'w') as f:
    f.write(test_file_content)

# Test the analysis
analyzer = TestAnalyzer()
analysis_result = analyzer.analyze_single_file(test_file_path)

print("✅ analyze_single_file method added and tested")
print(f"\nAnalysis result for {test_file_path}:")
print(f"  • Success: {'error' not in analysis_result}")
if 'error' not in analysis_result:
    print(f"  • Functions found: {len(analysis_result.get('functions', []))}")
    print(f"  • Classes found: {len(analysis_result.get('classes', []))}")
    print(f"  • Imports: {analysis_result.get('imports', [])}")
    print(f"  • Behaviors extracted: {len(analysis_result.get('behaviors', []))}")
    print(f"  • Quality score: {analysis_result.get('coverage_analysis', {}).get('quality_score', 'N/A')}")
else:
    print(f"  • Error: {analysis_result['error']}")

# Cleanup
test_file_path.unlink()

print("\nMethod breakdown:")
print("  • Log the analysis start")
print("  • Step 1: Parse test file using AST (mock_parse_test_file)")
print("  • Check for parsing errors and return early if found")
print("  • Step 2: Use AI summarizer for semantic analysis")
print("  • Return comprehensive analysis results")
print("  • Combines structural parsing with AI understanding")
```

```python
# Cell 5: analyze_directory Method - Batch Processing
def add_analyze_directory_method(analyzer_class):
    def analyze_directory(self, test_dir: Path, output_dir: Path) -> List[Dict[str, Any]]:
        """
        Analyze all test files in a directory and generate comprehensive reports.
        
        This method recursively scans a directory for test files, analyzes each one
        using both AST parsing and AI summarization, and generates detailed reports.
        """
        logger.info(f"Analyzing test directory: {test_dir}")
        
        if not test_dir.exists():
            raise FileNotFoundError(f"Test directory not found: {test_dir}")
        
        # Create output directory
        output_dir.mkdir(parents=True, exist_ok=True)
        
        # Find all test files
        test_files = self._find_test_files(test_dir)
        
        if not test_files:
            logger.warning(f"No test files found in {test_dir}")
            return []
        
        # Analyze each test file
        results = []
        for test_file in test_files:
            try:
                result = self.analyze_single_file(test_file)
                results.append(result)
                
                # Save individual file report
                self._save_file_report(test_file, result, output_dir)
                
            except Exception as e:
                logger.error(f"Error analyzing {test_file}: {e}")
                results.append({
                    'file_path': str(test_file),
                    'error': str(e)
                })
        
        # Generate aggregate report
        self._generate_aggregate_report(results, output_dir)
        
        return results
    
    analyzer_class.analyze_directory = analyze_directory

# Add the method to our analyzer
add_analyze_directory_method(TestAnalyzer)

print("✅ analyze_directory method added to TestAnalyzer")
print("\nMethod breakdown:")
print("  • test_dir: Path - Directory containing test files")
print("  • output_dir: Path - Where analysis reports will be written")
print("  • Validate test directory exists")
print("  • Create output directory structure")
print("  • Find all test files recursively")
print("  • Analyze each file individually")
print("  • Save individual file reports")
print("  • Generate aggregate summary report")
print("  • Return list of all analysis results")
```

```python
# Cell 6: Helper Methods for Directory Analysis
def add_directory_helper_methods(analyzer_class):
    def _find_test_files(self, test_dir: Path) -> List[Path]:
        """Find all test files in the directory recursively."""
        test_files = []
        
        for file_path in test_dir.rglob("*.py"):
            # Check if it's a test file
            if (file_path.name.startswith("test_") or 
                "test" in file_path.name.lower() or
                file_path.parent.name.lower() == "tests"):
                test_files.append(file_path)
        
        return sorted(test_files)
    
    def _save_file_report(self, test_file: Path, result: Dict[str, Any], output_dir: Path):
        """Save analysis report for a single test file."""
        # Create reports subdirectory
        reports_dir = output_dir / "reports"
        reports_dir.mkdir(exist_ok=True)
        
        # Generate report filename
        report_name = f"{test_file.stem}_analysis.json"
        report_path = reports_dir / report_name
        
        # Save the analysis result
        with open(report_path, 'w') as f:
            json.dump(result, f, indent=2, default=str)
    
    def _generate_aggregate_report(self, results: List[Dict[str, Any]], output_dir: Path):
        """Generate aggregate analysis report for all test files."""
        # Calculate aggregate statistics
        total_files = len(results)
        successful_analyses = len([r for r in results if 'error' not in r])
        failed_analyses = total_files - successful_analyses
        
        total_functions = sum(len(r.get('functions', [])) for r in results if 'error' not in r)
        total_behaviors = sum(len(r.get('behaviors', [])) for r in results if 'error' not in r)
        
        # Calculate average quality score
        quality_scores = [
            r.get('coverage_analysis', {}).get('quality_score', 0) 
            for r in results if 'error' not in r
        ]
        avg_quality = sum(quality_scores) / len(quality_scores) if quality_scores else 0
        
        aggregate_data = {
            'summary': {
                'total_files': total_files,
                'successful_analyses': successful_analyses,
                'failed_analyses': failed_analyses,
                'total_functions': total_functions,
                'total_behaviors': total_behaviors,
                'average_quality_score': avg_quality
            },
            'files': results
        }
        
        # Save aggregate report
        aggregate_path = output_dir / "aggregate_analysis.json"
        with open(aggregate_path, 'w') as f:
            json.dump(aggregate_data, f, indent=2, default=str)
    
    analyzer_class._find_test_files = _find_test_files
    analyzer_class._save_file_report = _save_file_report
    analyzer_class._generate_aggregate_report = _generate_aggregate_report

# Add the helper methods to our analyzer
add_directory_helper_methods(TestAnalyzer)

print("✅ Directory analysis helper methods added to TestAnalyzer")
print("\nHelper methods:")
print("  • _find_test_files - Recursively finds test files using naming patterns")
print("  • _save_file_report - Saves individual analysis results as JSON")
print("  • _generate_aggregate_report - Creates summary statistics for all files")
print("\nTest file detection patterns:")
print("  • Files starting with 'test_'")
print("  • Files containing 'test' in name")
print("  • Files in directories named 'tests'")
print("\nAggregate statistics calculated:")
print("  • Total files and success/failure counts")
print("  • Total functions and behaviors extracted")
print("  • Average quality score across all files")
```

```python
# Cell 7: Quality Assessment Methods
def add_quality_methods(analyzer_class):
    def assess_test_quality(self, analysis_result: Dict[str, Any]) -> Dict[str, Any]:
        """Assess the quality of a test file based on analysis results."""
        if 'error' in analysis_result:
            return {
                'overall_score': 0.0,
                'issues': [f"Analysis failed: {analysis_result['error']}"],
                'recommendations': ['Fix analysis errors first']
            }
        
        score = 0.0
        issues = []
        recommendations = []
        
        # Check function count
        functions = analysis_result.get('functions', [])
        if len(functions) == 0:
            issues.append("No test functions found")
        elif len(functions) < 3:
            issues.append("Low test function count")
            recommendations.append("Add more test functions")
        else:
            score += 0.3
        
        # Check docstrings
        functions_with_docs = sum(1 for f in functions if f.get('docstring'))
        if functions_with_docs == 0:
            issues.append("No function docstrings found")
            recommendations.append("Add docstrings to test functions")
        elif functions_with_docs / len(functions) < 0.8:
            issues.append("Some functions missing docstrings")
            score += 0.1
        else:
            score += 0.2
        
        # Check assertions
        functions_with_assertions = sum(1 for f in functions if f.get('assertions'))
        if functions_with_assertions == 0:
            issues.append("No assertions found in test functions")
            recommendations.append("Add assertions to validate test outcomes")
        else:
            score += 0.3
        
        # Check AI-extracted behaviors
        behaviors = analysis_result.get('behaviors', [])
        if len(behaviors) == 0:
            issues.append("No clear test behaviors identified")
        else:
            avg_confidence = sum(b.get('confidence', 0) for b in behaviors) / len(behaviors)
            score += 0.2 * avg_confidence
        
        return {
            'overall_score': min(score, 1.0),
            'issues': issues,
            'recommendations': recommendations,
            'breakdown': {
                'function_coverage': min(len(functions) / 5.0, 1.0) * 0.3,
                'documentation_quality': (functions_with_docs / len(functions)) * 0.2 if functions else 0,
                'assertion_coverage': (functions_with_assertions / len(functions)) * 0.3 if functions else 0,
                'behavior_clarity': avg_confidence * 0.2 if behaviors else 0
            }
        }
    
    analyzer_class.assess_test_quality = assess_test_quality

# Add the quality assessment method to our analyzer
add_quality_methods(TestAnalyzer)

# Test quality assessment
analyzer = TestAnalyzer()

# Create test analysis result
test_analysis = {
    'file_path': 'test_example.py',
    'functions': [
        {
            'name': 'test_with_docstring',
            'docstring': 'Test with documentation',
            'assertions': ['assert True']
        },
        {
            'name': 'test_without_docstring',
            'assertions': ['assert False']
        }
    ],
    'behaviors': [
        {'confidence': 0.8, 'behavior': 'Validates functionality'}
    ]
}

quality_result = analyzer.assess_test_quality(test_analysis)

print("✅ assess_test_quality method added and tested")
print(f"\nQuality assessment results:")
print(f"  • Overall score: {quality_result['overall_score']:.2f}")
print(f"  • Issues: {len(quality_result['issues'])}")
for issue in quality_result['issues']:
    print(f"    - {issue}")
print(f"  • Recommendations: {len(quality_result['recommendations'])}")
for rec in quality_result['recommendations']:
    print(f"    - {rec}")

print("\nMethod breakdown:")
print("  • Checks function count and coverage")
print("  • Evaluates documentation quality (docstrings)")
print("  • Assesses assertion coverage")
print("  • Analyzes AI-extracted behavior clarity")
print("  • Calculates weighted overall quality score")
print("  • Provides specific issues and improvement recommendations")
```
