```python
# Cell 3: CodeGenerator Class - Constructor
class CodeGenerator:
    """
    Main generator class that orchestrates AI-powered code generation from OpenSpec specifications.
    
    This class serves as the primary interface for converting OpenSpec YAML specifications
    into complete, working code implementations. It coordinates multiple AI models and
    generation strategies to produce source code, tests, documentation, and configuration
    files based on specification requirements.
    """
    
    def __init__(self, config_path: Optional[Path] = None):
        """
        Initialize the CodeGenerator with configuration.
        
        Args:
            config_path: Optional path to configuration file
        """
        self.config = self._load_config(config_path)
        self.spec_parser = MockSpecParser()
        self.ollama_client = MockOllamaClient({})
        self.templates = MockPromptTemplates()
        
        # Setup logging
        logging.basicConfig(level=logging.INFO)

# Test the constructor
generator = CodeGenerator()
print(f"✅ CodeGenerator created: {type(generator)}")
print(f"Config loaded: {type(generator.config)}")
print(f"Spec parser: {type(generator.spec_parser)}")
print(f"Ollama client: {type(generator.ollama_client)}")
print(f"Templates: {type(generator.templates)}")

print("\nConstructor breakdown:")
print("  • self.config = self._load_config(config_path) - Load configuration")
print("  • self.spec_parser = MockSpecParser() - Initialize specification parser")
print("  • self.ollama_client = MockOllamaClient({}) - Initialize AI client")
print("  • self.templates = MockPromptTemplates() - Initialize prompt templates")
print("  • logging.basicConfig(level=logging.INFO) - Setup logging")
```

```python
# Cell 4: _sanitize_filename Method - File System Safety
def add_sanitize_filename_method(generator_class):
    def _sanitize_filename(self, name: str) -> str:
        """
        Convert a name to a consistent, filesystem-safe filename format.
        
        This private method ensures that generated filenames are consistent across
        different platforms and filesystems by replacing problematic characters
        with safe alternatives.
        """
        # Always use underscores for consistency
        return name.replace('-', '_').replace(' ', '_')
    
    generator_class._sanitize_filename = _sanitize_filename

# Add the method to our generator
add_sanitize_filename_method(CodeGenerator)

# Test the method
generator = CodeGenerator()
test_names = [
    "my-api-spec",
    "test file name",
    "complex_name-v2.0",
    "User Authentication Service"
]

print("✅ _sanitize_filename method added and tested")
print("\nFilename sanitization examples:")
for name in test_names:
    sanitized = generator._sanitize_filename(name)
    print(f"  • '{name}' → '{sanitized}'")

print("\nMethod breakdown:")
print("  • name.replace('-', '_') - Replace hyphens with underscores")
print("  • .replace(' ', '_') - Replace spaces with underscores")
print("  • Returns filesystem-safe filename")
print("  • Ensures consistency across different platforms")
```

```python
# Cell 5: _load_config Method - Configuration Management
def add_load_config_method(generator_class):
    def _load_config(self, config_path: Optional[Path]) -> Dict[str, Any]:
        """
        Load generator configuration from a YAML file with fallback to default.
        
        This private method handles configuration loading for the code generator,
        supporting both custom configuration files and a default configuration.
        """
        # For demonstration, return a default configuration
        if config_path is None:
            return {
                'model': {
                    'model': 'qwen2.5-coder:7b',
                    'base_url': 'http://localhost:11434',
                    'temperature': 0.1,
                    'max_tokens': 4000
                },
                'generation': {
                    'overwrite': False,
                    'retry_attempts': 3
                }
            }
        else:
            # In real implementation, would load from file
            return {'model': {}, 'generation': {}}
    
    generator_class._load_config = _load_config

# Add the method to our generator
add_load_config_method(CodeGenerator)

# Test the method
generator = CodeGenerator()
default_config = generator._load_config(None)

print("✅ _load_config method added and tested")
print("\nDefault configuration structure:")
print(f"  • Model: {default_config['model']['model']}")
print(f"  • Base URL: {default_config['model']['base_url']}")
print(f"  • Temperature: {default_config['model']['temperature']}")
print(f"  • Max tokens: {default_config['model']['max_tokens']}")
print(f"  • Overwrite: {default_config['generation']['overwrite']}")
print(f"  • Retry attempts: {default_config['generation']['retry_attempts']}")

print("\nMethod breakdown:")
print("  • config_path: Optional[Path] - Path to custom config file")
print("  • Returns Dict[str, Any] - Configuration dictionary")
print("  • Default config includes model settings and generation options")
print("  • In real implementation: loads YAML from file")
```

```python
# Cell 6: generate_from_spec Method - Main Entry Point
def add_generate_from_spec_method(generator_class):
    def generate_from_spec(self, spec_path: Path, output_dir: Optional[Path] = None) -> GenerationResult:
        """
        Generate complete code implementation from an OpenSpec specification file.
        
        This is the main entry point for code generation. It processes an OpenSpec
        YAML file and generates a complete project structure.
        """
        start_time = time.time()
        files_generated = []
        errors = []
        warnings = []
        
        try:
            # Parse the specification
            spec = self.spec_parser.parse_file(spec_path)
            
            # Set up output directory
            if output_dir is None:
                output_dir = Path("generated_functions")
            
            output_dir.mkdir(exist_ok=True)
            
            # Generate files for each requirement
            for requirement in spec.requirements:
                try:
                    # Generate source code
                    source_file = self._generate_source_code(requirement, output_dir)
                    if source_file:
                        files_generated.append(source_file)
                    
                    # Generate test file
                    test_file = self._generate_test_file(requirement, output_dir)
                    if test_file:
                        files_generated.append(test_file)
                        
                except Exception as e:
                    errors.append(f"Error generating {requirement.id}: {str(e)}")
            
            # Generate documentation
            try:
                doc_file = self._generate_documentation(spec, output_dir)
                if doc_file:
                    files_generated.append(doc_file)
            except Exception as e:
                warnings.append(f"Could not generate documentation: {str(e)}")
            
            execution_time = time.time() - start_time
            success = len(errors) == 0
            
            return GenerationResult(
                success=success,
                files_generated=files_generated,
                errors=errors,
                warnings=warnings,
                execution_time=execution_time
            )
            
        except Exception as e:
            execution_time = time.time() - start_time
            return GenerationResult(
                success=False,
                files_generated=files_generated,
                errors=[f"Generation failed: {str(e)}"],
                warnings=warnings,
                execution_time=execution_time
            )
    
    generator_class.generate_from_spec = generate_from_spec

# Add the method to our generator
add_generate_from_spec_method(CodeGenerator)

print("✅ generate_from_spec method added to CodeGenerator")
print("\nMethod breakdown:")
print("  • start_time = time.time() - Track execution start")
print("  • Parse specification using spec_parser")
print("  • Create output directory if not specified")
print("  • Loop through requirements and generate files")
print("  • Generate source code and test files for each requirement")
print("  • Generate documentation for the complete specification")
print("  • Return GenerationResult with all outcomes")
```

```python
# Cell 7: Helper Methods for File Generation
def add_generation_helper_methods(generator_class):
    def _generate_source_code(self, requirement, output_dir: Path) -> Optional[str]:
        """Generate source code file for a requirement."""
        try:
            # Get prompt for this requirement
            prompt = self.templates.get_generation_prompt(requirement)
            
            # Generate code using AI
            generated_code = self.ollama_client.generate(prompt)
            
            # Create filename
            filename = self._sanitize_filename(requirement.title.lower()) + ".py"
            file_path = output_dir / filename
            
            # Write to file
            with open(file_path, 'w') as f:
                f.write(generated_code)
            
            return str(file_path)
            
        except Exception as e:
            logging.error(f"Error generating source code for {requirement.id}: {e}")
            return None
    
    def _generate_test_file(self, requirement, output_dir: Path) -> Optional[str]:
        """Generate test file for a requirement."""
        try:
            # Create test filename
            filename = "test_" + self._sanitize_filename(requirement.title.lower()) + ".py"
            file_path = output_dir / filename
            
            # Generate basic test structure
            test_code = f"""import pytest
from {self._sanitize_filename(requirement.title.lower())} import *

def test_{self._sanitize_filename(requirement.title.lower())}():
    """Test {requirement.title}"""
    # TODO: Implement test cases
    assert True
"""
            
            with open(file_path, 'w') as f:
                f.write(test_code)
            
            return str(file_path)
            
        except Exception as e:
            logging.error(f"Error generating test file for {requirement.id}: {e}")
            return None
    
    def _generate_documentation(self, spec, output_dir: Path) -> Optional[str]:
        """Generate documentation for the specification."""
        try:
            doc_path = output_dir / "README.md"
            
            doc_content = f"""# {spec.title}

Generated from OpenSpec specification: {spec.spec_id}

## Requirements

"""
            
            for req in spec.requirements:
                doc_content += f"\n### {req.id}: {req.title}\n\n{req.description}\n"
            
            with open(doc_path, 'w') as f:
                f.write(doc_content)
            
            return str(doc_path)
            
        except Exception as e:
            logging.error(f"Error generating documentation: {e}")
            return None
    
    generator_class._generate_source_code = _generate_source_code
    generator_class._generate_test_file = _generate_test_file
    generator_class._generate_documentation = _generate_documentation

# Add the helper methods to our generator
add_generation_helper_methods(CodeGenerator)

print("✅ Generation helper methods added to CodeGenerator")
print("\nHelper methods:")
print("  • _generate_source_code - Creates Python files from requirements")
print("  • _generate_test_file - Creates test files for each requirement")
print("  • _generate_documentation - Creates README.md with specification info")
print("\nEach method handles:")
print("  • File naming and sanitization")
print("  • Content generation using AI or templates")
print("  • File writing with proper error handling")
print("  • Logging for debugging")
```
